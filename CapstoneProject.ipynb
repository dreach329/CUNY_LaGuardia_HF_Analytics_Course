```python
#rename columns for alignment
df2 = df2.rename(columns={'age': 'Age',
                            'user_id': 'User_ID',
                             'daily_screen_time_hours': 'Screen_Time_Hours',
                             'social_media_hours': 'Social_Media_Usage_Hours',
                             'sleep_hours': 'Sleep_Hours',
                             'exercise_minutes': 'Physical_Activity_Hours',
                             'study_work_hours': 'Study_Work_Hours',
                             'productivity_score': 'Productivity_Score',
                             'stress_level': 'Stress_Level',
                             'platform': 'Platform'})
```

```python
#Age Breakdown
combined['Age'].value_counts()
```

```python
#Visualization Age Breakdown

plt.figure(figsize=(10, 6))
sns.countplot(data=combined, x='Age', color='Teal')
plt.title('Age Distribution', size=15)
plt.show()
```

```python
#additional age visualization = particpants concentrated in the 25-48 group
plt.boxplot(combined['Age'])
plt.title('Age Distribution')
plt.show()
```

```python
#Avg Screen time by Age
avg_screen_time = combined.groupby('Age')['Screen_Time_Hours'].mean()

plt.figure(figsize=(10,6))
sns.barplot(x=avg_screen_time.index, y=avg_screen_time.values, color = 'purple')
plt.title('Average Screen Time by Age')
plt.xlabel('Age')
plt.ylabel('Screen Time (Hours)')
plt.xticks(rotation=90)
plt.show()
```

```python
#Creating Age Bands
combined["Age_Band"] = pd.cut(
    combined["Age"],
    bins=[0, 18, 25, 35, 45, 60, 100],
    labels=["Under 18", "18-25", "26-35", "36-45", "46-60", "60+"])
```

```python
#Stress Level Breakdown
combined['Stress_Level'].value_counts()
```

```python
#Stess Level visualization
sns.countplot(data=combined, x='Stress_Level', )
plt.title('Stress Level Distribution', size=15)
plt.show()
```

```python
#Arrange by Age and Stress_Level
combined.groupby(['Age', 'Stress_Level']).value_counts()
```

```python
# Calculating the proportion of each stress level by age group
stress_level_age = combined.groupby(['Age', 'Stress_Level']).size().unstack().fillna(0)

# Normalizing to get proportions
stress_level_age = stress_level_age.div(stress_level_age.sum(axis=1), axis=0)

# Plotting a stacked bar chart
plt.figure(figsize=(15,8))
stress_level_age.plot(kind='bar', stacked=True)
plt.title('Stress Level Distribution by Age Group')
plt.xlabel('Age')
plt.ylabel('Proportion of Stress Levels')
plt.xticks(rotation=90)
plt.legend(title='Stress Level', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
#Stacked view of stress by age bands
combined.groupby(["Age_Band", "Stress_Level"]).size().unstack().plot(
    kind="bar",
    stacked=True
)

import matplotlib.pyplot as plt

plt.title("Stress Levels by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Count")

plt.show()
```

```python
combined["Stress_Score"] = combined["Stress_Level"].map({
    "Low": 1,
    "Medium": 2,
    "High": 3
})
```

```python
corr = combined[[
    "Age",
    "Social_Media_Usage_Hours",
    "Screen_Time_Hours",
    "Sleep_Hours",
    "Physical_Activity_Hours",
    "Stress_Score"
]].corr()
```

```python
#how does each variable correlate with stress
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Lifestyle Factors and Stress")

plt.show()
```
